## Overview

Tokenization is the process of breaking text into smaller units called tokens. These tokens can be words, sentences, or even characters. Tokenization is a crucial first step in most NLP pipelines, as it converts raw text into a format that can be processed by machine learning algorithms.

- Regular expression tokenization

- Tweet and social media tokenization

## Why Tokenization Matters

Tokenization:
- Converts unstructured text into structured data
- Enables counting and statistical analysis
- Facilitates feature extraction for machine learning
- Helps identify patterns and relationships in text
- Is the foundation for more advanced NLP tasks

## Import Libraries

```python
import nltk
import re
from nltk.tokenize import (
    word_tokenize,
    sent_tokenize,
    wordpunct_tokenize,
    TreebankWordTokenizer,
    WhitespaceTokenizer,
    RegexpTokenizer,
    TweetTokenizer
)

# Download required NLTK data
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

print("Libraries imported successfully!")
```

## 1. Basic Word Tokenization

The most common tokenization is splitting text into words.

```python
# English example
english_text = "My name is John. I love Natural Language Processing!"
english_tokens = word_tokenize(english_text)

print("English Text:")
print(f"  Original: {english_text}")
print(f"  Tokens: {english_tokens}")
print(f"  Number of tokens: {len(english_tokens)}")
```

```python
# Vietnamese example
vietnamese_text = "Tên tôi là John. Tôi yêu xử lý ngôn ngữ tự nhiên!"
vietnamese_tokens = word_tokenize(vietnamese_text)

print("Vietnamese Text:")
print(f"  Original: {vietnamese_text}")
print(f"  Tokens: {vietnamese_tokens}")
print(f"  Number of tokens: {len(vietnamese_tokens)}")
```

## 2. Sentence Tokenization

Breaking text into sentences is useful for many NLP tasks.

```python
# English paragraph
english_paragraph = """
Hello! My name is John. I love programming. 
Natural Language Processing is fascinating. Let's learn together!
"""

english_sentences = sent_tokenize(english_paragraph)

print("English Paragraph:")
print(english_paragraph)
print("\nSentences:")
for i, sentence in enumerate(english_sentences, 1):
    print(f"{i}. {sentence.strip()}")
```

```python
# Vietnamese paragraph
vietnamese_paragraph = """
Xin chào! Tên tôi là John. Tôi yêu lập trình.
Xử lý ngôn ngữ tự nhiên rất hấp dẫn. Hãy cùng học nhé!
"""

vietnamese_sentences = sent_tokenize(vietnamese_paragraph)

print("Vietnamese Paragraph:")
print(vietnamese_paragraph)
print("\nSentences:")
for i, sentence in enumerate(vietnamese_sentences, 1):
    print(f"{i}. {sentence.strip()}")
```

## 3. Different Tokenization Methods

NLTK provides several tokenization methods, each with different behaviors.

```python
text = "Don't worry! It's going to be great. Email: user@example.com"

print(f"Original text: {text}\n")

# 1. word_tokenize (Penn Treebank tokenizer)
tokens1 = word_tokenize(text)
print(f"word_tokenize: {tokens1}\n")

# 2. wordpunct_tokenize (splits on punctuation)
tokens2 = wordpunct_tokenize(text)
print(f"wordpunct_tokenize: {tokens2}\n")

# 3. TreebankWordTokenizer
treebank_tokenizer = TreebankWordTokenizer()
tokens3 = treebank_tokenizer.tokenize(text)
print(f"TreebankWordTokenizer: {tokens3}\n")

# 4. WhitespaceTokenizer
whitespace_tokenizer = WhitespaceTokenizer()
tokens4 = whitespace_tokenizer.tokenize(text)
print(f"WhitespaceTokenizer: {tokens4}")
```

## 4. Regular Expression Tokenization

Create custom tokenizers using regular expressions.

```python
# Tokenize only alphabetic words
alpha_tokenizer = RegexpTokenizer(r'[A-Za-z]+')
text = "Hello! I have 3 apples and 5 oranges."

print(f"Original: {text}")
print(f"Alphabetic tokens only: {alpha_tokenizer.tokenize(text)}")
```

```python
# Tokenize words and numbers separately
word_num_tokenizer = RegexpTokenizer(r'\w+')
text = "I bought 3 apples for $5.50 in 2024."

print(f"Original: {text}")
print(f"Words and numbers: {word_num_tokenizer.tokenize(text)}")
```

```python
# Tokenize email addresses
email_tokenizer = RegexpTokenizer(r'[\w.+-]+@[\w-]+\.[\w.-]+')
text = "Contact me at john@example.com or jane.doe@company.co.uk"

print(f"Original: {text}")
print(f"Email addresses: {email_tokenizer.tokenize(text)}")
```

## 5. Tweet and Social Media Tokenization

TweetTokenizer handles hashtags, mentions, and emoticons.

```python
tweet_tokenizer = TweetTokenizer()

tweet = "@john: I love #NLP and #MachineLearning! 😊 Check this out: https://example.com"

print(f"Original tweet: {tweet}")
print(f"\nTokens: {tweet_tokenizer.tokenize(tweet)}")
```

```python
# TweetTokenizer with options
tweet_tokenizer_custom = TweetTokenizer(
    preserve_case=False,  # Convert to lowercase
    reduce_len=True,      # Reduce repeated characters
    strip_handles=False    # Keep @ mentions
)

tweet = "@user Helloooooo!!! This is AWESOME!!! :) :) :)"

print(f"Original tweet: {tweet}")
print(f"Tokens (custom): {tweet_tokenizer_custom.tokenize(tweet)}")
```

## 6. Handling Edge Cases

Let's see how different tokenizers handle challenging cases.

```python
# Contractions
text_contractions = "I can't believe it's already 2024! We're learning NLP."

print(f"Text with contractions: {text_contractions}")
print(f"word_tokenize: {word_tokenize(text_contractions)}")
print(f"wordpunct_tokenize: {wordpunct_tokenize(text_contractions)}")
```

```python
# Hyphenated words
text_hyphen = "State-of-the-art machine learning models are mind-blowing!"

print(f"Text with hyphens: {text_hyphen}")
print(f"word_tokenize: {word_tokenize(text_hyphen)}")
print(f"WhitespaceTokenizer: {whitespace_tokenizer.tokenize(text_hyphen)}")
```

```python
# Numbers and decimals
text_numbers = "The price is $19.99 and the tax is 8.5%."

print(f"Text with numbers: {text_numbers}")
print(f"word_tokenize: {word_tokenize(text_numbers)}")
print(f"RegexpTokenizer (\\w+): {word_num_tokenizer.tokenize(text_numbers)}")
```

## 7. Custom Tokenizer Function

Create your own tokenizer tailored to specific needs.

```python
def custom_tokenizer(text, 
                     lowercase=True,
                     remove_punct=False,
                     keep_numbers=True):
    """
    Custom tokenizer with configurable options.
    
    Parameters:
    -----------
    text : str
        Input text to tokenize
    lowercase : bool
        Convert to lowercase
    remove_punct : bool
        Remove punctuation tokens
    keep_numbers : bool
        Keep number tokens
    
    Returns:
    --------
    list : List of tokens
    """
    # Tokenize
    tokens = word_tokenize(text)
    
    # Lowercase
    if lowercase:
        tokens = [token.lower() for token in tokens]
    
    # Remove punctuation
    if remove_punct:
        tokens = [token for token in tokens if token.isalnum()]
    
    # Remove numbers
    if not keep_numbers:
        tokens = [token for token in tokens if not token.isdigit()]
    
    return tokens

# Test custom tokenizer
text = "Hello! I have 3 apples and 5 ORANGES."

print(f"Original: {text}\n")
print(f"Default: {custom_tokenizer(text)}")
print(f"Remove punct: {custom_tokenizer(text, remove_punct=True)}")
print(f"Remove punct & numbers: {custom_tokenizer(text, remove_punct=True, keep_numbers=False)}")
```

## 8. Comparing Tokenization Methods

Let's compare different tokenization methods on the same text.

```python
text = """Hello! I'm John, a data scientist. 
I work on state-of-the-art NLP models. 
Contact me at john@example.com or visit https://example.com.
#MachineLearning #AI"""

print("Original Text:")
print(text)
print("\n" + "="*70 + "\n")

# Different tokenization methods
methods = [
    ("word_tokenize", word_tokenize(text)),
    ("wordpunct_tokenize", wordpunct_tokenize(text)),
    ("WhitespaceTokenizer", whitespace_tokenizer.tokenize(text)),
    ("TweetTokenizer", tweet_tokenizer.tokenize(text)),
]

for method_name, tokens in methods:
    print(f"{method_name}:")
    print(f"  Tokens: {tokens[:10]}...")  # Show first 10 tokens
    print(f"  Total tokens: {len(tokens)}")
    print()
```

## 9. Practical Exercise: Bilingual Text Analysis

Compare tokenization for English and Vietnamese parallel texts.

```python
# Parallel English-Vietnamese text
english = "My name is John. I love programming and machine learning."
vietnamese = "Tên tôi là John. Tôi yêu lập trình và học máy."

# Tokenize both
english_tokens = word_tokenize(english)
vietnamese_tokens = word_tokenize(vietnamese)

print("English → Vietnamese Tokenization Comparison:")
print("="*70)
print(f"\nEnglish: {english}")
print(f"Tokens: {english_tokens}")
print(f"Token count: {len(english_tokens)}")

print(f"\nVietnamese: {vietnamese}")
print(f"Tokens: {vietnamese_tokens}")
print(f"Token count: {len(vietnamese_tokens)}")

print(f"\nDifference in token count: {abs(len(english_tokens) - len(vietnamese_tokens))}")
```

```python
# More examples
translation_pairs = [
    ("Hello", "Xin chào"),
    ("Thank you", "Cảm ơn"),
    ("How are you?", "Bạn khỏe không?"),
    ("I love programming", "Tôi yêu lập trình"),
]

print("Translation Pairs - Token Analysis:")
print("="*70)

for english, vietnamese in translation_pairs:
    en_tokens = word_tokenize(english)
    vi_tokens = word_tokenize(vietnamese)
    
    print(f"\nEnglish: {english}")
    print(f"  Tokens: {en_tokens} ({len(en_tokens)} tokens)")
    print(f"Vietnamese: {vietnamese}")
    print(f"  Tokens: {vi_tokens} ({len(vi_tokens)} tokens)")
```

## 10. Performance Considerations

```python
import time

# Generate a large text sample
large_text = " ".join(["This is a sample sentence for performance testing."] * 1000)

print(f"Text length: {len(large_text)} characters\n")

# Test different tokenizers
tokenizers = [
    ("word_tokenize", lambda t: word_tokenize(t)),
    ("wordpunct_tokenize", lambda t: wordpunct_tokenize(t)),
    ("WhitespaceTokenizer", lambda t: whitespace_tokenizer.tokenize(t)),
    ("RegexpTokenizer", lambda t: word_num_tokenizer.tokenize(t)),
]

print("Performance Comparison:")
print("="*70)

for name, tokenizer_func in tokenizers:
    start_time = time.time()
    tokens = tokenizer_func(large_text)
    end_time = time.time()
    
    print(f"{name}:")
    print(f"  Time: {(end_time - start_time)*1000:.2f} ms")
    print(f"  Tokens: {len(tokens)}")
    print()
```

## Key Takeaways

1. **Choose the Right Tokenizer**: Different tokenizers are suitable for different tasks
   - `word_tokenize`: General-purpose, handles contractions well
   - `TweetTokenizer`: Best for social media text
   - `RegexpTokenizer`: Flexible, customizable with regex patterns
   - `WhitespaceTokenizer`: Simple and fast

2. **Language Matters**: Tokenization behavior can differ across languages
   - Vietnamese uses spaces differently than English
   - Some languages (Chinese, Japanese) don't use spaces at all
   - Consider language-specific tokenizers for better accuracy

3. **Edge Cases**: Be aware of how tokenizers handle:
   - Contractions (can't, won't)
   - Hyphenated words
   - URLs and email addresses
   - Numbers and punctuation
   - Emoticons and special characters

4. **Performance**: Simple tokenizers (whitespace, regex) are faster than sophisticated ones

5. **Context Dependent**: Tokenization strategy should match your downstream task
