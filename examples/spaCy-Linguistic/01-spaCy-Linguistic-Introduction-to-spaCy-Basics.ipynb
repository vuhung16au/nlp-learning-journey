## 2. Basic Text Processing

### Creating a Doc Object

The `Doc` object is spaCy's core data structure. It contains all the linguistic annotations for a text.


```python
# Create a Doc object by processing text with our nlp pipeline
text = "Hello world! This is a sample text for spaCy processing."
doc = nlp(text)

# The Doc object contains all the processed information
print("Original text:", doc.text)
print("Number of tokens:", len(doc))
print("Number of sentences:", len(list(doc.sents)))
```

### Understanding Tokens

Tokens are the individual units of text (words, punctuation, etc.). Let's explore them:


```python
# Iterate through tokens
print("Token analysis:")
print("-" * 50)
for token in doc:
    print(f"Token: '{token.text}'")
    print(f"  - Is alpha: {token.is_alpha}")
    print(f"  - Is punctuation: {token.is_punct}")
    print(f"  - Is space: {token.is_space}")
    print(f"  - Is stop word: {token.is_stop}")
    print()
```

### Accessing Basic Attributes

Let's explore some basic token attributes:


```python
# Let's work with a more interesting text
text = "The quick brown fox jumps over the lazy dog."
doc = nlp(text)

print("Token attributes:")
print("-" * 60)
for token in doc:
    print(f"'{token.text:10}' | POS: {token.pos_:8} | Tag: {token.tag_:8} | Lemma: {token.lemma_:10}")
```

## 3. Part-of-Speech Tagging

### What is POS Tagging?

Part-of-speech tagging assigns grammatical categories (noun, verb, adjective, etc.) to each word in a sentence. This is fundamental for understanding the structure and meaning of text.


```python
# Practical examples of POS tagging
text = "I love programming in Python and building amazing applications!"
doc = nlp(text)

print("POS Tagging Example:")
print("=" * 50)
for token in doc:
    if not token.is_space:  # Skip whitespace tokens
        print(f"'{token.text:12}' -> {token.pos_:8} ({token.tag_:8})")
```

### Common POS Tags

spaCy uses the Universal Dependencies (UD) tagset. Here are some common tags:


```python
# Let's analyze different types of words
text = "The beautiful cat gracefully jumped over the tall fence yesterday."
doc = nlp(text)

# Group tokens by POS tag
pos_groups = {}
for token in doc:
    if not token.is_space and not token.is_punct:
        pos = token.pos_
        if pos not in pos_groups:
            pos_groups[pos] = []
        pos_groups[pos].append(token.text)

print("Words grouped by Part of Speech:")
print("=" * 40)
for pos, words in pos_groups.items():
    print(f"{pos:8}: {', '.join(words)}")
```

## 4. Lemmatization

### What is Lemmatization?

Lemmatization reduces words to their base or dictionary form (lemma). For example:
- "running" → "run"
- "better" → "good" 
- "mice" → "mouse"

This is different from stemming, which just chops off word endings.


```python
# Comparing word forms vs lemmas
text = "The cats are running quickly and jumping over fences."
doc = nlp(text)

print("Word Forms vs Lemmas:")
print("=" * 30)
for token in doc:
    if not token.is_space and not token.is_punct:
        print(f"'{token.text:10}' -> '{token.lemma_:10}'")
```

### Practical Applications

Lemmatization is useful for:
- Text normalization
- Information retrieval
- Reducing vocabulary size
- Improving text analysis accuracy


```python
# Example: Text normalization using lemmatization
text = "I am running, jumping, and swimming every day to stay healthy."
doc = nlp(text)

# Extract lemmatized words (excluding stop words and punctuation)
lemmatized_words = []
for token in doc:
    if not token.is_stop and not token.is_punct and not token.is_space:
        lemmatized_words.append(token.lemma_.lower())

print("Original text:", text)
print("Lemmatized words:", lemmatized_words)
print("Unique lemmatized words:", set(lemmatized_words))
```
