```python
import spacy
from spacy import displacy

# Load the English language model
nlp = spacy.load("en_core_web_sm")
```

## 1. Dependency Parsing Basics

### What is Dependency Parsing?

Dependency parsing analyzes the grammatical structure of a sentence by identifying relationships between words. Each word has a "head" (the word it depends on) and a "dependency label" (the type of relationship).


```python
# Basic dependency parsing example
text = "The quick brown fox jumps over the lazy dog."
doc = nlp(text)

print("Dependency Parsing Example:")
print("=" * 40)
for token in doc:
    print(f"'{token.text:8}' -> {token.dep_:12} -> '{token.head.text:8}'")
```

### Understanding Dependency Labels

Common dependency labels include:
- **nsubj**: Nominal subject
- **dobj**: Direct object
- **amod**: Adjectival modifier
- **det**: Determiner
- **prep**: Prepositional modifier
- **pobj**: Object of preposition


```python
# Analyzing dependency relationships in detail
text = "The intelligent student reads interesting books in the library."
doc = nlp(text)

print("Detailed Dependency Analysis:")
print("=" * 50)
for token in doc:
    if not token.is_space and not token.is_punct:
        print(f"Token: '{token.text}'")
        print(f"  - POS: {token.pos_}")
        print(f"  - Dependency: {token.dep_}")
        print(f"  - Head: '{token.head.text}'")
        print(f"  - Head POS: {token.head.pos_}")
        print()
```

## 2. Navigating the Parse Tree

### Accessing Token Dependencies

We can navigate the parse tree by following relationships between tokens:


```python
# Finding the root of the sentence
text = "The cat sat on the mat."
doc = nlp(text)

# Find the root token (usually the main verb)
root = [token for token in doc if token.head == token][0]
print(f"Sentence: {text}")
print(f"Root token: '{root.text}' (POS: {root.pos_})")

# Find all children of the root
print(f"\nChildren of '{root.text}':")
for child in root.children:
    print(f"  - '{child.text}' ({child.dep_})")
```

### Finding Roots, Children, and Ancestors

We can traverse the parse tree in different directions:


```python
# Practical tree traversal
text = "The beautiful red car drives quickly down the street."
doc = nlp(text)

print("Parse Tree Traversal:")
print("=" * 30)

# Find all subjects
subjects = [token for token in doc if token.dep_ == "nsubj"]
print("Subjects:")
for subj in subjects:
    print(f"  - '{subj.text}' (head: '{subj.head.text}')")

# Find all objects
objects = [token for token in doc if token.dep_ in ["dobj", "pobj"]]
print("\nObjects:")
for obj in objects:
    print(f"  - '{obj.text}' (head: '{obj.head.text}')")

# Find all adjectives
adjectives = [token for token in doc if token.dep_ == "amod"]
print("\nAdjectives:")
for adj in adjectives:
    print(f"  - '{adj.text}' (modifies: '{adj.head.text}')")
```

### Practical Tree Traversal

Let's see how to navigate relationships in more complex sentences:


```python
# Advanced tree traversal example
text = "The student who studies hard will pass the exam."
doc = nlp(text)

print("Advanced Parse Tree Analysis:")
print("=" * 40)

# Find the main verb
main_verb = [token for token in doc if token.head == token][0]
print(f"Main verb: '{main_verb.text}'")

# Find all noun phrases
noun_phrases = []
for token in doc:
    if token.pos_ == "NOUN":
        # Collect the noun and its modifiers
        phrase = [token.text]
        for child in token.children:
            if child.dep_ in ["det", "amod", "compound"]:
                phrase.append(child.text)
        noun_phrases.append(" ".join(phrase))

print(f"\nNoun phrases: {noun_phrases}")

# Find relative clauses
relative_pronouns = [token for token in doc if token.dep_ == "nsubj" and token.head.dep_ == "relcl"]
print(f"\nRelative clauses: {[token.text for token in relative_pronouns]}")
```

## 3. Noun Chunks

### Extracting Noun Phrases

Noun chunks are "base noun phrases" - flat phrases that have a noun as their head. They're useful for information extraction and text analysis.


```python
# Extracting noun chunks
text = "The quick brown fox jumps over the lazy dog in the beautiful garden."
doc = nlp(text)

print("Noun Chunks:")
print("=" * 20)
for chunk in doc.noun_chunks:
    print(f"'{chunk.text}' (root: '{chunk.root.text}')")
    print(f"  - Start: {chunk.start}, End: {chunk.end}")
    print(f"  - Root POS: {chunk.root.pos_}")
    print()
```

### Use Cases and Applications

Noun chunks are useful for:
- Information extraction
- Named entity recognition
- Text summarization
- Question answering systems


```python
# Practical example: Information extraction
text = "Apple Inc. is a technology company founded by Steve Jobs in California."
doc = nlp(text)

print("Information Extraction Example:")
print("=" * 35)

# Extract all noun chunks
print("All noun chunks:")
for chunk in doc.noun_chunks:
    print(f"  - {chunk.text}")

# Extract specific types of information
companies = [chunk.text for chunk in doc.noun_chunks if "company" in chunk.text.lower()]
people = [chunk.text for chunk in doc.noun_chunks if chunk.root.ent_type_ == "PERSON"]
places = [chunk.text for chunk in doc.noun_chunks if chunk.root.ent_type_ == "GPE"]

print(f"\nCompanies: {companies}")
print(f"People: {people}")
print(f"Places: {places}")
```

## 4. Visualizing Dependencies

### Using displacy for Visualization

spaCy's `displacy` module provides excellent visualization tools for dependency parsing:


```python
# Basic dependency visualization
text = "The quick brown fox jumps over the lazy dog."
doc = nlp(text)

# Create dependency visualization
displacy.render(doc, style="dep", jupyter=True)
```

### Interpreting Dependency Diagrams

The visualization shows:
- **Arrows**: Point from dependent to head
- **Labels**: Show the relationship type
- **Colors**: Different colors for different relationship types
