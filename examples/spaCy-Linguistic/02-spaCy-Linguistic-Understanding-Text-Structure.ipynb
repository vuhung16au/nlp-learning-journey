```python
import spacy

# Load the English language model
nlp = spacy.load("en_core_web_sm")
```

## 1. Tokenization Fundamentals

### How spaCy Tokenizes Text

Tokenization is the process of splitting text into meaningful segments. spaCy uses a sophisticated algorithm that balances accuracy with efficiency.


```python
# Basic tokenization example
text = "Hello world! This is a test."
doc = nlp(text)

print("Tokenization example:")
print("=" * 30)
for i, token in enumerate(doc):
    print(f"Token {i}: '{token.text}'")
    print(f"  Start: {token.idx}, End: {token.idx + len(token.text)}")
    print(f"  Whitespace: '{token.whitespace_}'")
    print()
```

### Special Cases in Tokenization

spaCy handles various special cases intelligently:


```python
# Special cases: contractions, URLs, emails, etc.
special_cases = [
    "I can't believe it!",
    "Visit https://spacy.io for more info",
    "Contact me at john.doe@example.com",
    "It's 3:30 PM on Monday",
    "The price is $19.99",
    "Dr. Smith will see you now"
]

print("Special Cases in Tokenization:")
print("=" * 40)

for text in special_cases:
    doc = nlp(text)
    print(f"\nText: {text}")
    print("Tokens:", [token.text for token in doc])
```

### Understanding Token Properties

Each token has various properties that help us understand its role in the text:


```python
# Analyzing token properties
text = "The quick brown fox jumps over the lazy dog."
doc = nlp(text)

print("Token Properties Analysis:")
print("=" * 50)
for token in doc:
    print(f"Token: '{token.text}'")
    print(f"  - is_alpha: {token.is_alpha}")
    print(f"  - is_digit: {token.is_digit}")
    print(f"  - is_punct: {token.is_punct}")
    print(f"  - is_space: {token.is_space}")
    print(f"  - is_stop: {token.is_stop}")
    print(f"  - like_email: {token.like_email}")
    print(f"  - like_url: {token.like_url}")
    print()
```

## 2. Sentence Segmentation

### Splitting Text into Sentences

Sentence segmentation is the process of identifying sentence boundaries in text. This is crucial for many NLP tasks.


```python
# Basic sentence segmentation
text = "Hello world! This is the first sentence. This is the second sentence. And this is the third one."
doc = nlp(text)

print("Sentence Segmentation:")
print("=" * 30)
for i, sent in enumerate(doc.sents):
    print(f"Sentence {i+1}: {sent.text.strip()}")
    print(f"  Number of tokens: {len(sent)}")
    print()
```

### Edge Cases and Challenges

Sentence segmentation can be tricky with abbreviations, decimal numbers, and other special cases:


```python
# Challenging sentence segmentation cases
challenging_texts = [
    "Dr. Smith went to the U.S.A. He returned at 3:30 p.m.",
    "The price is $19.99. That's a great deal!",
    "Visit www.example.com for more info. Call us at (555) 123-4567.",
    "I love Python! It's amazing. What do you think?"
]

print("Challenging Sentence Segmentation:")
print("=" * 40)

for text in challenging_texts:
    doc = nlp(text)
    print(f"\nText: {text}")
    sentences = [sent.text.strip() for sent in doc.sents]
    print(f"Sentences ({len(sentences)}): {sentences}")
```

## 3. Morphological Analysis

### Understanding Morphological Features

Morphological analysis examines the internal structure of words, including tense, number, gender, and other grammatical features.


```python
# Exploring morphological features
text = "The cats are running quickly and jumping over fences."
doc = nlp(text)

print("Morphological Analysis:")
print("=" * 30)
for token in doc:
    if not token.is_space and not token.is_punct:
        print(f"Word: '{token.text}'")
        print(f"  Lemma: {token.lemma_}")
        print(f"  Morphology: {token.morph}")
        print(f"  POS: {token.pos_}")
        print()
```

### Exploring Word Properties

Let's examine different types of morphological features:


```python
# Analyzing different word types
text = "She walked to the store yesterday. The children were playing happily."
doc = nlp(text)

print("Detailed Morphological Analysis:")
print("=" * 40)

for token in doc:
    if not token.is_space and not token.is_punct:
        print(f"'{token.text}' -> {token.lemma_}")
        
        # Check for specific morphological features
        morph_features = []
        if token.morph.get("Tense"):
            morph_features.append(f"Tense: {token.morph.get('Tense')[0]}")
        if token.morph.get("Number"):
            morph_features.append(f"Number: {token.morph.get('Number')[0]}")
        if token.morph.get("Person"):
            morph_features.append(f"Person: {token.morph.get('Person')[0]}")
        if token.morph.get("Gender"):
            morph_features.append(f"Gender: {token.morph.get('Gender')[0]}")
        if token.morph.get("Degree"):
            morph_features.append(f"Degree: {token.morph.get('Degree')[0]}")
            
        if morph_features:
            print(f"  Features: {', '.join(morph_features)}")
        print()
```

### Practical Examples

Let's see how morphological analysis helps in real-world scenarios:


```python
# Example: Finding all past tense verbs
text = "I walked to school yesterday. She ran quickly. They jumped over the fence."
doc = nlp(text)

print("Finding Past Tense Verbs:")
print("=" * 30)

past_tense_verbs = []
for token in doc:
    if (token.pos_ == "VERB" and 
        token.morph.get("Tense") and 
        token.morph.get("Tense")[0] == "Past"):
        past_tense_verbs.append(token.text)

print(f"Past tense verbs found: {past_tense_verbs}")

# Example: Finding plural nouns
print("\nFinding Plural Nouns:")
print("=" * 25)

plural_nouns = []
for token in doc:
    if (token.pos_ == "NOUN" and 
        token.morph.get("Number") and 
        token.morph.get("Number")[0] == "Plur"):
        plural_nouns.append(token.text)

print(f"Plural nouns found: {plural_nouns}")
```

### Exercise 3: Morphological Feature Extraction
Extract specific morphological features from text.


```python
# Your solution here
def extract_morphological_features(text):
    """
    Extract morphological features from text
    """
    doc = nlp(text)
    
    features = {
        'past_tense_verbs': [],
        'plural_nouns': [],
        'comparative_adjectives': [],
        'superlative_adjectives': []
    }
    
    # TODO: Complete the feature extraction
    for token in doc:
        if not token.is_space and not token.is_punct:
            # Extract past tense verbs
            if (token.pos_ == "VERB" and 
                token.morph.get("Tense") and 
                token.morph.get("Tense")[0] == "Past"):
                features['past_tense_verbs'].append(token.text)
            
            # Extract plural nouns
            if (token.pos_ == "NOUN" and 
                token.morph.get("Number") and 
                token.morph.get("Number")[0] == "Plur"):
                features['plural_nouns'].append(token.text)
            
            # Extract comparative adjectives
            if (token.pos_ == "ADJ" and 
                token.morph.get("Degree") and 
                token.morph.get("Degree")[0] == "Cmp"):
                features['comparative_adjectives'].append(token.text)
            
            # Extract superlative adjectives
            if (token.pos_ == "ADJ" and 
                token.morph.get("Degree") and 
                token.morph.get("Degree")[0] == "Sup"):
                features['superlative_adjectives'].append(token.text)
    
    return features

# Test the function
test_text = "The children ran faster than the adults. The biggest dog was the happiest."
result = extract_morphological_features(test_text)

print("Morphological Features:")
print("=" * 30)
for feature_type, words in result.items():
    print(f"{feature_type}: {words}")
```
